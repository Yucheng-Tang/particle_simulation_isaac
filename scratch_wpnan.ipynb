### A kernel to check nan value

```python
import warp as wp
from warp import render
from typing import Any
```

```python
@wp.kernel
def check_none(
    input: wp.array(dtype=Any),
    output: wp.array(dtype=bool)
):
    if not input:
        output[0] = True
        wp.printf('None found')
    else:
        output[0] = False
        wp.printf('No None found')
```

Set output as an array, because a kernel can only modify an array

```python
input = wp.array([None])
output = wp.array([False])

wp.launch(
    kernel=check_none,
    inputs=[input, output],
    dim=1
)
print(f'output is{output[0]}')
```
